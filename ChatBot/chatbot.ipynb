```python
from langchain_ollama import ChatOllama
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from typing import TypedDict, Annotated
from langchain_core.messages import SystemMessage, HumanMessage, BaseMessage
from pydantic import BaseModel
from langgraph.checkpoint.memory import MemorySaver # saves the state in-memory (RAM)
```

```python
class ChatState(TypedDict):

    messages: Annotated[list[BaseMessage], add_messages]
```

```python
llm = ChatOllama(model="mistral", temperature=0.7)
def chat_node(state: ChatState) -> ChatState:
    messages = state["messages"]
    res = llm.invoke(messages)
    return {"messages": [res]}
```

```python
    checkpointer = MemorySaver()
    graph = StateGraph(ChatState)

    graph.add_node("chat_node", chat_node)
    graph.add_edge(START, "chat_node")
    graph.add_edge("chat_node", END)

    workflow = graph.compile(checkpointer=checkpointer)
```

```python
from IPython.display import Image
Image(workflow.get_graph().draw_mermaid_png())
```

```python
query = {"messages": [
    HumanMessage(content="What is the capital of France?")
]}
workflow.invoke(query)
```

```python
thread_id = '1'

while True:
    query = input("You: ")
    if query.strip().lower() in ['exit', 'quit']:
        print("Exiting chat.")
        break

    query_state = {"messages": [HumanMessage(content=query)]}
    config = {'configurable' : {'thread_id': thread_id}}
    response_state = workflow.invoke(query_state, config=config)#adding this config will make sure that the state is stored in the checkpointer with the thread_id as the key, and when we invoke the workflow again with the same thread_id, it will fetch the state from the checkpointer and use it as the input to the chat_node, thus making the chatbot remember the previous messages in the conversation.
    response_messages = response_state["messages"][-1].content
    print("You: " + query)
    print(f"Bot: {response_messages}") 
# Here the chatbot will not remember anything even when we are adding everything to the state. This is because we are invoking the workflow in every single iteration, and because of that the state is fresh every time. To fix this we need to use the concept of persistence.
```
