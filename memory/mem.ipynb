```python
from langgraph.store.memory import InMemoryStore
```

```python
namespace = ("user", "u1")
store = InMemoryStore()
```

```python
store.put(namespace, "1", {"data" : "user likes pizza."})
store.put(namespace, "2", {"data" : "user's favorite color is blue."})

store.put(("user", "u2"), "1", {"data" : "user likes pasta."})
store.put(("user", "u2"), "2", {"data" : "user's favorite color is red."})
```

```python
print(store.get(namespace, "1"))
items = store.search(namespace)
for item in items:
    print(item)
    print("--" * 10)
```
